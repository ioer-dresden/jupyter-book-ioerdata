# Restricted file access

This guide explains how to register on ioerDATA and how to access restricted files. Restricted files are datasets with limited access, primarily to comply with IOER FDZ regulations and protect sensitive data.

Files may be restricted if they:
- Originate from licensed statistical sources,
- could enable indirect identification when combined with other data,
- are subject to usage agreements with data providers.

```{admonition} Registration required for downloads
:class: important 
To comply with IOER FDZ regulations, **registered users can download datasets**. Descriptive files (e.g., `README.md`, `LIESMICH.md`, `Datadescription.pdf`, `thumbnails`) are publicly accessible.

If you don't ioerDATA account yet, register for a free account on the [ioerDATA platform](https://data.fdz.ioer.de/dataverse/ioer_fdz)([sign‑up page](https://data.fdz.ioer.de/dataverseuser.xhtml?editMode=CREATE&redirectPage=%2Floginpage.xhtml)).  
```

You may also find restricted files containing sensitive information (e.g., personal survey data) that are only accessible via stricter user agreements with IOER FDZ.

## Access restricted files 

To work with restricted files from ioerDATA, you must have a user account.
You can access restricted files in two ways, depending on your preferences:

### 1. Request access on the dataset page

All datasets can be downloaded directly on the website:  

1. Log in ioerDATA.
2. Navigate to the selected dataset or replication package page.
3. Locate the restricted dataset/file.
4. Click `Access Dataset` or `Access File` and follow the instructions.

**Note:** Restricted files are marked with a small lock icon. This lock will be:
- Red if permission is denied.
- Green if permission is granted.


 ```{figure} ../resources/Screenshot_Access_Data.png
:name: Access files

Access restricted files on ioerDATA webpage.
```


### 2. Accessing your personal API token

ioerDATA uses personal API tokens to authenticate requests and identify the calling user. These tokens are unique to each user. 

Personal API tokens allow you to download restricted datasets programmatically from ioerDATA using your own user account.

**Steps to access your API token:**

1. Log in ioerDATA.
2. In your ioerDATA account, click on the top right corner on your user name and select *API Token*.
