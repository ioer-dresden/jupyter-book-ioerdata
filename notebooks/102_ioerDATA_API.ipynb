# Working with ioerDATA API

An API (Application Programming Interface) allows users to automatically request and retrieve data from a data platform using a personal API token. Instead of manually downloading files from a website, an API enables direct access to datasets within your analysis workflow. Working with the ioerDATA API increases the reproducibility of your work, as APIs make your data analysis workflow transparent and easy to repeat and reproduce, even for other users. This follows the FAIR principles.

ioerDATA uses the API provided by *The Dataverse Project*. For more information on how to use your API token with the Dataverse APIs, check out the [API Guide](https://guides.dataverse.org/en/6.10.1/api/).
