# How to work with this book
Each chapter begins with a main section introducing key concepts for working with ioerDATA datasets, designed for all readers. These chapters will familiarize you with the process and prepare you to use our datasets in your own projects.
The content is divided into two main parts. The **first part** provides an overview of how to:

- Access research data from ioerDATA using Python.
- Explore downloaded datasets or replication packages.
- Understand files, variables, and metadata.
- Analyze and visualize spatial research data.
- Make your workflow reproducible.
  
To make this book more fun, we’ve created in the **second part** the Urban Data Investigator 🕵️ and several missions!

## Urban Data Investigator 🕵️ 

Follow the same research workflow as in part 1 and learn how to use ioerDATA datasets as a sequence of missions:

🔎 Find → 📥 Retrieve → 🧠 Understand → 📊 Analyse → 🗺️ Visualise → 🔁 Reproduce

Each mission builds on what you discovered in the previous one.

## What will we work with? 

This book uses two real ioerDATA datasets. 

- **Part 1**: This part uses the data from the [replication package for climate regulation in cities](https://data.fdz.ioer.de/dataset.xhtml?persistentId=doi:10.71830/AFW3N3), published by *Syrbe et al. 2025*. With this dataset, we will explore urban green infrastructure and related indicators while learning how research data moves from a repository into a reproducible Python workflow.

- **Part 2**: The urban data investigator 🕵️ focuses on data from [replication package for localized assessment of urban forest structures with 3D indicators](https://data.fdz.ioer.de/dataset.xhtml?persistentId=doi:10.71830/CDAXYF), published by *Münziger et al. 2026*.
