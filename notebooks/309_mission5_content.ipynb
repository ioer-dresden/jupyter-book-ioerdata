It works — but the figure tells the reader very little.

A useful scientific visualisation should make it clear:

>- **what** is being shown,
>- **what the axes mean**, and

# 🌳 Compare canopy area and volume

Canopy area tells us **how much space is covered**, while canopy volume captures its **3D structure**.

A scatter plot can help us see how these two indicators relate.

````python
plt.scatter(
    berlin["chm_area"],
    berlin["chm_volume"],
    alpha=0.4
)

plt.title("Canopy area and canopy volume in Berlin")
plt.xlabel("Canopy area")
plt.ylabel("Canopy volume")

plt.show()
````

# 🎨 Make the evidence clearer

>Small design choices can make a figure easier to read without adding unnecessary decoration.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(
    berlin["chm_area"],
    berlin["chm_volume"],
    alpha=0.35
)

ax.set_title("Relationship between canopy area and volume in Berlin")
ax.set_xlabel("Canopy area")
ax.set_ylabel("Canopy volume")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
````
