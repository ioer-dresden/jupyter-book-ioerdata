## Setting up your Computing Environment

This book is available as **Jupyter Notebooks**, which are interactive documents that combine explanations, code, and outputs in one place. It can be made interactive using Binder from [Jupyter4NFDI](https://base4nfdi.de/projects/jupyter4nfdi) or Google Colab. You can run our notebooks entirely in your browser without installing anything locally. Click the 🚀 icon in the top menu bar of any notebook page and choosing `Jupyter4NFDI` or `Colab`.

<img src="../resources/binder.png" width="50%" style="display: inline-block">

### Option A: Jupyter4NFDI

The Binder from Jupyter4NFDI is a great solution for German institutions or users with existing ORCID/Google accounts.

- Select Jupyter4NFDI from the 🚀 menu.
- Authenticate via the Helmholtz AAI (Tip: You can use your ORCID or Google account if you don’t have an institutional login).
- The environment loads with JupyterLab, Git, and all packages pre-installed.

### Option B: Google Colab

Google Colab is best for international users who prefer a quick Google login.

- Select Colab from the 🚀 menu.
- Our notebooks contain a setup cell that automatically syncs the repository, data paths, and installs required packages on the fly. Both platforms function identically!

> SStarting Binder or Google Colab may take a moment. The page will appear divided into cells. Please have a look below at the [Quick-start guide on how to use Jupyter](https://stag.ioerdata.training.fdz.ioer.info/notebooks/006_how_to_work.html#quick-start-guide-on-how-to-use-jupyter) if you’d like to get familiar with the functions of Jupyter.


### Option C: Carto-Lab Docker 
If you are a more experienced user, you can use **Carto-Lab Docker**, which is also officially supported by the IOER FDZ. Carto-Lab Docker is a reproducible environment for Spatial Data Science. It provides a versioned, pre-configured JupyterLab server in a Docker container, designed to make transparent and reproducible spatial analysis accessible to everyone.

This option is useful when you want to reproduce the full software environment used for this training.

Find out more about [CartoLab Docker](https://cartolab.fdz.ioer.info/) and how to use it.

### Option D: Work locally

If you already an expert in Python or Jupyter, you can clone the complete training repository and work with the notebooks locally.

View the [training repository](https://gitlab.hrz.tu-chemnitz.de/ioer/fdz/training/jupyter-book-ioerdata/-/tree/staging?ref_type=heads).
