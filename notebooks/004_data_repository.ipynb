# Research data repository

A **research data repository** is a digital platform for the storage, publication, and long-term access of research data. It is recommended and important for researchers to upload their data to such repositories to ensure that valuable information and findings are not lost and remain accessible to others!

Research data repositories follow the *FAIR principles*. The FAIR principles are guidelines that make data Findable, Accessible, Interoperable, and Reusable. This ensures data is well-organized, machine-readable, and optimized for reuse across disciplines.

Repositories provide well-structured metadata and method documentation, helping others to understand the datasets. They also ensure data provenance, documenting the origin and processing history of datasets and ensuring reproducibility of the findings.


## ioerDATA

ioerDATA is the institutional research data repository of the [Leibniz Institute of Ecological Urban and Regional Development (IOER)](https://www.ioer.de/en/). It is managed and curated by the institute's [Research Data Centre (IOER FDZ)](https://ioer-fdz.de/en/).




IOER researchers can publish their research data in the ioerDATA repository, which uses the open-source software [The Dataverse Project](https://dataverse.org/). All datasets published in ioerDATA follow the FAIR principles and are assigned a Digital Object Identifier (DOI) for persistent citation and discovery.

There are two types of datasets in ioerDATA: **Curated IOER Data** and **Data Supporting Publications**: 

::::{tab-set}
:::{tab-item} Curated IOER Data
:sync: tab1
> Strategic, high-demand foundational datasets with broad research applications. 
*   **Purpose:** Long-term access to flagship IOER datasets for integration into ongoing research.
*   **Content:** Well-documented, reviewed and quality-checked. Includes rich metadata and documentation explaining the broader context, methodology, and potential applications. Time series are provided for some datasets.
:::
:::{tab-item} Data Supporting Publications
:sync: tab2
> Datasets and **replication packages** linked to a specific publication (e.g., journal article, report, dissertation).
*   **Purpose:** Provide everything needed for someone else to repeat an analysis and obtain the same results reported in a paper, including figures, statistics, and any other output shared in the paper. 
*   **Content:** Raw or processed **datasets** used in the analysis, **scripts** for processing, analysis, and visualization (including dependencies and versions), and rich **documentation (README)** with step-by-step instructions for reproducing results.
:::

::::



## Metadata

Metadata is often described as *data about data*. The metadata in ioerDATA follows a defined, standardized format, which makes it uniform and machine-readable for all published datasets. All metadata are descriptive and publicly accessible. In ioerDATA there are different kinds of metadata provided: Citation Metadata, Geospatial Metadata, Computational Workflow Metadata, and File-level Metadata.

::::{tab-set}
:::{tab-item} Citation Metadata
:sync: tab1
Citation Metadata provides the essential information needed to uniquely identify and properly cite a dataset. ioerDATA provides a series of citation metadata fields that need to be filled in and are hence provided. Here are some examples:  

Examples:

- **Persistent Identifier:** The datasets's unique persistent identifier, either a DOI or Handle. 
- **Publication Date:** The publication data of a dataset. 
- **Title:** The main title of a dataset.
- **Author:** The entity, e.g. a person or organization, that created the Dataset.
:::
:::{tab-item} Geospatial Metadata
:sync: tab2
IOER works with geospatial data, scince then, ioerDATA provides domain specific metadata for geospatial datasets. It includes: 

- **Geographic Coverage:** Information on the geographic coverage of the data. Includes the total geographic scope of the data. 
- **Geographic Unit:** Lowest level of geographic aggregation covered by the dataset, e.g., village, county, region.
- **Geographic Bounding Box**: The fundamental geometric description for any dataset that models geography is the geographic bounding box. It describes the minimum box, defined by west and east longitudes and north and south latitudes, which includes the largest geographic extent of the dataset's geographic coverage. This element is used in the first pass of a coordinate-based search. Inclusion of this element in the codeblock is recommended, but is required if the bound polygon box is included. 
:::
:::{tab-item} Computational Workflow Metadata
:sync: tab3
Describes the steps, tools, and processes used to generate, process, or analyze the dataset.

- **Workflow Type:** The kind of Computational Workflow, which is designed to compose and execute a series of computational or data manipulation steps in a scientific application. 
:::
::::

**File-level Metadata**

File-level Metadata describe the individual files within a dataset, including details such as:
- File name,
- File type/format (e.g., CSV, PDF, Shapefile),
- File size,
- Access status (public or restricted),
- Description.
