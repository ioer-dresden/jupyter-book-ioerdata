# Research data repository

A **research data repository** is a digital platform for the storage, publication, and long-term access of research data. It is recommended and important for researchers to upload their data to such repositories to ensure that valuable information and findings are not lost and remain accessible to others!

Research data repositories follow the *FAIR principles*. The FAIR principles are guidelines that make data Findable, Accessible, Interoperable, and Reusable. This ensures data is well-organized, machine-readable, and optimized for reuse across disciplines.

Repositories provide well-structured metadata and method documentation, helping others to understand the datasets. They also ensure data provenance, documenting the origin and processing history of datasets and ensuring reproducibility of the findings.


## ioerDATA

ioerDATA is the institutional research‑data publication repository of the [Leibniz Institute of Ecological Urban and Regional Development (IOER)](https://www.ioer.de/en/). It is managed and curated by the institute's [Research Data Centre (IOER FDZ)](https://ioer-fdz.de/en/).

IOER researchers can publish their research data in the ioerDATA repository, which uses the open-source software [The Dataverse Project](https://dataverse.org/). All datasets published in ioerDATA follow the FAIR principles and are assigned a Digital Object Identifier (DOI) for persistent citation and discovery.

There are two types of datasets in ioerDATA: **Curated IOER Data** and **Data Supporting Publications**

### Curated IOER Data
Curated IOER Data are strategic, high-demand foundational datasets with broad research applications. 

- **Purpose:**
  Long-term access to **flagship IOER datasets** for integration into ongoing research.

- **Content:**
  Well-documented, reviewed and quality-checked. Includes rich metadata and documentation explaining the broader context, methodology, and potential applications. Time series are provided for some datasets.

- **Process:**
  Rigorous internal review, curation, and quality assurance with the **IOER-FDZ team**. Long-term commitment to quality, usability, and maintenance (e.g., websites, APIs). 

### Data Supporting Publications
Datasets and replication packages linked to a specific publication (e.g., journal article, report, dissertation). In ioerDATA, several replication packages are already provided. A good replication package provide everything needed for someone else to repeat an analysis and obtain the same results reported in a paper, including  figures, statistics, and any other output shared in the paper.

- **Purpose:**
  Transparency and **verifiability** of research findings.

- **Content:**
  - **Data:** Raw or processed datasets used in the analysis.
  - **Code:** Scripts for processing, analysis, and visualization (including dependencies and versions).
  - **Documentation (README):**
    Step-by-step instructions for reproducing results.

- **Ideal:**
  Self-contained **replication packages** (data + code + documentation) to ensure full reproducibility.
