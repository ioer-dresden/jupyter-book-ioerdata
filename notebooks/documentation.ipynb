# Creating training notebooks

Training content follows a progressive approach:

# Learn → Run → Extend

**Learn**  
Introduce the concept and explain why it matters.

**Run**  
Provide a small, copy-paste-ready example.

**Extend**  
Encourage learners to modify the example or apply it elsewhere.

# Recommended notebook structure

A training notebook should normally contain:

1. **Questions** — What problem are we solving?
2. **Learning objectives** — What will the learner be able to do?
3. **Context** — Why does this matter?
4. **Code** — Small executable steps
5. **Output** — Data, figures, or maps
6. **Interpretation** — What does the result mean?
7. **Takeaway** — What should the learner remember?
8. **Extension** — What could they try next?

> 💡 Not every notebook needs every section. Keep the structure proportional to the lesson.

# Writing guidelines

Write for a learner who may be using Python or spatial data for the first time.

**Prefer**

>- short explanations
>- one concept at a time
>- meaningful variable names
>- reusable code
>- visible intermediate results
>- interpretation after analyses
>- descriptive figure titles
>- links to data sources

**Avoid**

>- unexplained technical terms
>- large code blocks doing many things at once
><- unnecessary dependencies
>- hidden setup steps
>- repeated explanations
>- manual modification of generated outputs

# Hiding Notebook Cells

To hide the source code of a notebook cell in the published Jupyter Book:

1. Select the notebook cell.
2. Open **View → Property Inspector**.
3. Under **Cell Metadata**, add the following tag:

```text
remove-input
```

The cell will still be executed during notebook execution, but its input will not be displayed in the published book.

```{note}
Hidden cells are still executed when the notebook runs. The `remove-input` tag only hides the source code in the published Jupyter Book while keeping the generated output visible.
```

# Figures and visualisations

Figures should communicate a result, not simply demonstrate that plotting code works.

Whenever appropriate:

- use descriptive titles and labels
- include units
- explain what the reader should notice
- minimise unnecessary visual clutter
- include data-source attribution
- provide geographic context for maps
- make important figures downloadable

For reusable figures, keep a reference to the Matplotlib figure:

```python
fig, ax = plt.subplots()

gdf.plot(ax=ax)

ax.set_title("Descriptive title")

This makes later export straightforward:

````python
fig.savefig("figure.png", dpi=300, bbox_inches="tight")
fig.savefig("figure.pdf", bbox_inches="tight")
````

# Reproducibility

A learner should be able to understand **where an output came from and how to recreate it**.

Whenever possible:

**Data → Code → Analysis → Figure → Export**

Therefore:

- retrieve source data programmatically
- document data sources and versions
- keep processing steps visible
- reuse previously defined variables
- generate outputs from code
- avoid manual editing of datasets and figures
- document important assumptions
- keep notebooks executable from beginning to end

> 💡 If a result cannot be recreated from the documented workflow, treat it as a reproducibility issue.

# Build and test

Do not rely only on how a notebook looks inside JupyterLab.

After making changes:

>1. save the notebook
>2. restart and run it from beginning to end
>3. build the Jupyter Book
>4. inspect the rendered page
>5. test links and downloads
>6. check figures and hidden cells
>7. review the staging deployment

The generated `_build/` directory contains build output and should not be treated as source content.

# Pre-commit check

Before committing, ask:

> **Does it run? → Does it build? → Does it render? → Does it download?**

Check especially:

>- notebook execution
>- internal links
>- image paths
>- downloadable PNG/PDF files
>- `_toc.yml`
>- hidden-cell behaviour
>- light/dark-mode styling
>- browser-based notebook compatibility where relevant

# Debug systematically

When something breaks, identify **where** it breaks:

>**Notebook**
→ Does the code execute?

>**Build**
→ Does Jupyter Book build successfully?

>**Repository**
→ Are all required files tracked?

>**Deployment**
→ Did CI/CD complete successfully?

>**Browser**
→ Is the correct deployed version being displayed?

Fix the problem at the earliest stage where it appears.

# Troubleshooting

Most issues encountered while working with the training materials fall into one of the following categories.

```{Environment}

If notebooks fail to execute or packages cannot be imported, check that:

- the correct Python kernel (**worker_env**) is selected;
- all required packages are installed;
- your environment is up to date.
```

```{Git}

Common version control issues include:

- merge conflicts;
- untracked files;
- forgetting to update `_toc.yml` after adding or renaming notebooks.
```

```{Jupyter Book}

If the book does not build as expected, verify that:

- internal links are valid;
- referenced images exist;
- hidden cells are configured correctly;
- notebooks execute successfully from beginning to end.
```

```{Deployment}

If your changes are not visible after a successful pipeline:

1. Verify that the deployment has completed successfully.
2. Refresh your browser using **Ctrl + F5**.
3. Confirm that you are viewing the correct website (staging or production).
```
