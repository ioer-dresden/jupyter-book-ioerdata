The navigation of the published Jupyter Book is controlled by the `_toc.yml` file. Whenever a notebook is added, removed, or renamed, this file should also be updated.

```{tip}
Before creating a new notebook, check whether a similar example already exists. Reusing and extending existing content helps maintain consistency across the training materials.
```

# Development Environment

The training materials are developed and tested using **CartoLab** and **JupyterLab**.

For consistency across all notebooks, use the **worker_env** kernel whenever possible. This environment already includes the required scientific Python packages and avoids installing additional dependencies.

When developing locally, ensure that the required Python libraries are available before executing the notebooks.

```{seealso}
If you are new to CartoLab or would like to learn more about the platform, the following resources provide a good starting point:

- **CartoLab Quick Start Guide:**  
  https://cartolab.fdz.ioer.info/quickstart/

- **CartoLab Concept:**  
  https://cartolab.fdz.ioer.info/concept/
```

```{important}
To ensure that all notebooks execute consistently, use the **worker_env** kernel whenever possible. This environment is maintained with the required dependencies used throughout the training.
```

```{warning}
Always pull the latest changes from the remote repository before starting work. This reduces the likelihood of merge conflicts and ensures that you are working with the most recent version of the training materials.
```

# Hiding Notebook Cells

To hide the source code of a notebook cell in the published Jupyter Book:

1. Select the notebook cell.
2. Open **View → Property Inspector**.
3. Under **Cell Metadata**, add the following tag:

```text
remove-input
```

The cell will still be executed during notebook execution, but its input will not be displayed in the published book.

```{note}
Hidden cells are still executed when the notebook runs. The `remove-input` tag only hides the source code in the published Jupyter Book while keeping the generated output visible.
```

# Creating Training Materials

To keep the training consistent and beginner friendly, every notebook should follow a similar structure.

A recommended order is:

1. Learning objectives
2. Background information
3. Explanatory text
4. Executable Python code
5. Interpretation of the results
6. Key takeaways
7. Optional exercises or extensions

```{tip}
Aim to keep each notebook self-contained. A reader should be able to execute every notebook from beginning to end without requiring manual intervention or hidden setup steps.
```

# Writing Guidelines

When creating new content:

- Explain concepts before introducing code.
- Assume little or no prior experience.
- Keep explanations concise and practical.
- Provide copy-paste-ready code examples.
- Reuse variables where appropriate.
- Interpret figures and results instead of only generating them.
- Write notebooks that can be executed from start to finish without manual intervention.

```{important}
Remember that the primary audience includes beginners. Whenever introducing a new concept, explain *why* it is useful before showing *how* to implement it.
```

# Building the Jupyter Book

After making changes to notebooks or documentation:

1. Save all modified files.
2. Build the Jupyter Book.
3. Ensure that all notebooks execute successfully.
4. Check that figures, downloads, and internal links work correctly.
5. Review the generated pages before publishing.

```{tip}
Check both the notebook output and the rendered Jupyter Book page. Some formatting issues only become visible after the documentation has been built.
```

# Reproducibility

The training is designed to promote reproducible research.

Whenever possible:

- Download data directly from ioerDATA.
- Keep processing steps transparent.
- Generate figures programmatically.
- Avoid manual editing of outputs.
- Keep notebooks executable from beginning to end.
- Document assumptions and processing decisions.

```{note}
Whenever possible, avoid manually editing figures or datasets. Producing all outputs directly from the notebooks improves transparency and makes analyses easier to reproduce.
```

# Troubleshooting

Most issues encountered while working with the training materials fall into one of the following categories.

```{Environment}

If notebooks fail to execute or packages cannot be imported, check that:

- the correct Python kernel (**worker_env**) is selected;
- all required packages are installed;
- your environment is up to date.
```

```{Git}

Common version control issues include:

- merge conflicts;
- untracked files;
- forgetting to update `_toc.yml` after adding or renaming notebooks.
```

```{Jupyter Book}

If the book does not build as expected, verify that:

- internal links are valid;
- referenced images exist;
- hidden cells are configured correctly;
- notebooks execute successfully from beginning to end.
```

```{Deployment}

If your changes are not visible after a successful pipeline:

1. Verify that the deployment has completed successfully.
2. Refresh your browser using **Ctrl + F5**.
3. Confirm that you are viewing the correct website (staging or production).
```

```{warning}
If an issue persists after checking the documentation, verify whether a recent commit or dependency update may have introduced unexpected behaviour before making additional changes.
```

```{tip}
Small, focused commits with clear commit messages are easier to review and simplify collaboration within the project.
```

# Contributing

Contributions are welcome.

Before submitting changes:

- Test all notebooks from beginning to end.
- Verify that figures and downloads are generated correctly.
- Follow the existing writing style.
- Keep explanations beginner friendly.
- Use clear and descriptive commit messages.
- Update the documentation whenever necessary.

# Changelog

Significant updates to the training materials should be recorded here.

| Version | Date | Author | Description |
|----------|------|--------|-------------|
| 1.0 | YYYY-MM-DD | IOER Team | Initial release |
