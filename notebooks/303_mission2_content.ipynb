# 🗂️ What's inside a replication package?

A replication package is more than a dataset.

>It can contain **data, documentation, code, and outputs** that together help others understand and reproduce a study.

Before analysing anything, let's find out what our package contains.

# 📂 Step 1: Inspect the downloaded files

>Mission 1 stored the downloaded files in `data/g_raw`.

Let's list them and take our first look at the package.

````python
from pathlib import Path

# Folder containing files downloaded in Mission 1
data_folder = Path("data/g_raw")

# Collect all files in the folder
files = sorted(data_folder.glob("*"))

print("Files in replication package:\n")

for file in files:
    print(f"- {file.name}")
````

```{admonition} 🕵️ First clues
:class: note

Look at the filenames and extensions.

Can you already spot:

- 📄 documentation?
- 💻 executable code?
- 🗺️ spatial data?
- 📊 analysis outputs?

The file extension often gives us our first clue about a file's purpose.

# 🔎 Step 2: Identify the file types

>Instead of inspecting the filenames one by one, let's create a simple overview of the package.

````python
for file in files:
    print(f"{file.name:<80} {file.suffix}")
````

# 🧩 Read the package

Different file types play different roles in the research workflow.

For this package, some important examples are:

| File type | What it tells us |
|---|---|
| `.md` | Documentation |
| `.ipynb` | Reproducible analysis workflow |
| `.gpkg` | Spatial data |
| `.rds` | R data or analysis objects |
| `.png` | Visual output |
| `.html` | Static report |

Together, these files form the **replication package**.

```{tip}
Do not start with the largest or most complicated file.

A good first stop is often the **README** — it can explain the package structure, data sources, software requirements, and how the analysis should be reproduced.

# 📘 Step 3: Read the README File

>Many replication packages include a README file explaining the dataset structure and workflow.

````python
# Path to README file
readme_path = data_folder / "README.md"

# Check whether README exists before reading it
if readme_path.exists():
    with open(readme_path, "r", encoding="utf-8") as f:
        readme_text = f.read()

    print(readme_text[:1500])
else:
    print("README.md was not found in the downloaded files.")
````

```{admonition} 🧠 Think Like a Researcher
:class: tip

Good researchers rarely start with analysis immediately.

They first ask:

- What does the README explain?
- Which files are important?
- Which workflow was originally used?
- Which components are missing?
