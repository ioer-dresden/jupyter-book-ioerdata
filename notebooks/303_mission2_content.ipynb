### 📂 Step 1: Inspect the downloaded files

>Mission 1 stored the downloaded files in `data/g_raw`.

Let's list them and take our first look at the package.

````python
from pathlib import Path

# Folder containing files downloaded in Mission 1
data_folder = Path("data/g_raw")

# Collect all files in the folder
files = sorted(data_folder.glob("*"))

print("Files in replication package:\n")

for file in files:
    print(f"- {file.name}")
````

```{admonition} 🕵️ First clues
:class: note

Look at the filenames and extensions.

Can you already spot:

- 📄 documentation?
- 💻 executable code?

- 📊 analysis outputs?

The file extension often gives us our first clue about a file's purpose.

### 🔎 Step 2: Identify the file types

>Instead of inspecting the filenames one by one, let's create a simple overview of the package.

````python
for file in files:
    print(f"{file.name:<80} {file.suffix}")
````


#### 🧩 Decode the Evidence: Understanding File Types

Every file in your replication package has a purpose. As an **Urban Data Investigator**, your next task is to identify these files and understand their roles in the research workflow.

| File type | Role in the investigation |
|---|---|
| `.md` | 📄 Documentation explaining the dataset and research |
| `.ipynb` | 💻 Jupyter Notebook containing reproducible analysis code |
| `.gpkg` | 🗺️ GeoPackage containing spatial data |
| `.rds` | 📊 R data files or saved analysis objects |
| `.png` | 🖼️ Visual outputs such as charts and maps |
| `.html` | 📑 HTML reports or other web-based outputs |

Together, these files provide the evidence needed to **understand, verify, and reproduce the research**.

🔍 **Investigator's clue:** Knowing what each file contains is the first step towards reconstructing the research workflow.



```{tip} 🔍 Investigator's Tip
Before opening the largest or most complex file, look for the **README**.

Think of it as your investigation guide. It often contains essential clues about the package structure, data sources, software requirements, and steps needed to reproduce the research.
```

### 📘 Step 3: Read the README File

>Many replication packages include a README file explaining the dataset structure and workflow.

````python
# Path to README file
readme_path = data_folder / "README.md"

# Check whether README exists before reading it
if readme_path.exists():
    with open(readme_path, "r", encoding="utf-8") as f:
        readme_text = f.read()

    print(readme_text[:1500])
else:
    print("README.md was not found in the downloaded files.")
````

```{admonition} 🧠 Think Like a Researcher
:class: tip

Good researchers rarely start with analysis immediately.

They first ask:

- What does the README explain?
- Which files are important?
- Which workflow was originally used?
- Which components are missing?
```
