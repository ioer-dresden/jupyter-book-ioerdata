## 📂 Step 1: List Downloaded Files

Use this cell to inspect the files downloaded from Mission 1.

````python
from pathlib import Path

# Folder containing files downloaded in Mission 1
data_folder = Path("data/g_raw")

# Collect all files in the folder
files = sorted(data_folder.glob("*"))

print("Files in replication package:\n")

for file in files:
    print(f"- {file.name}")
````

## 📄 Step 2: Identify File Types

File extensions help us understand what role each file may play in the replication package.

````python
from collections import Counter

# Extract file extensions from the downloaded files
extensions = [file.suffix.lower() for file in files]

# Count how many times each extension appears
extension_counts = Counter(extensions)

print("File types found:\n")

for ext, count in extension_counts.items():
    label = ext if ext else "[no extension]"
    print(f"{label}: {count}")
````

```{admonition} 💡 Understanding File Types
:class: note

Different file types usually serve different purposes:

- `.md` → documentation  
- `.ipynb` → Jupyter notebooks / workflows  
- `.html` → exported reports or visual outputs  
- `.png` → figures or graphics  
- `.gpkg` → geospatial data  
- `.rds` → serialized R objects  

🧠 Understanding file structure is an important research skill.
```

## 📘 Step 3: Read the README File

Many replication packages include a README file explaining the dataset structure and workflow.

````python
# Path to README file
readme_path = data_folder / "README.md"

# Check whether README exists before reading it
if readme_path.exists():
    with open(readme_path, "r", encoding="utf-8") as f:
        readme_text = f.read()

    print(readme_text[:1500])
else:
    print("README.md was not found in the downloaded files.")
````
