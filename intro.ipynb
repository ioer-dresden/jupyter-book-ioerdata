Welcome! 👋

This hands-on training shows you how to work with research data from [ioerDATA](https://data.fdz.ioer.de/) using **Python and Jupyter notebooks**.

You will move step by step from finding a dataset to understanding, analysing, and visualising it.

<hr style="border: none; border-top: 8px solid #70a401;">

# 🔍 Questions

By working through this training, we will answer:

>- How can I access research data from ioerDATA using Python?
>- What is inside a downloaded dataset or replication package?
>- How can I understand its files, variables, and metadata?
>- How can I analyse and visualise spatial research data?
>- How can I make my workflow reproducible?

<hr style="border: none; border-top: 8px solid #70a401;">

# 🎯 Learning objectives

By the end of the training, you will be able to:

>- access datasets through the ioerDATA API
>- download research data with Python
>- inspect files and metadata
>- work with spatial data and GeoPackages
>- explore indicators and create visualisations
>- document a reproducible research workflow

<hr style="border: none; border-top: 8px solid #70a401;">

# What is ioerDATA?

[ioerDATA](https://data.fdz.ioer.de/) is the research data publication repository of the [Leibniz Institute of Ecological Urban and Regional Development (IOER)](https://www.ioer.de/en/).

It provides access to research datasets, code, documentation, and replication packages for spatial, environmental, and socio-economic research.

Throughout this training, we use real ioerDATA research data to learn by doing.

<hr style="border: none; border-top: 8px solid #70a401;">

# How to use this book

You can use the training at three levels:

>1. Learn

Read the explanations and follow the examples step by step.

**No prior Python experience is required.**

>2. Run

Copy and run the code examples to reproduce the analyses, maps, and visualisations.

>3. Extend

Modify the examples to explore other indicators, datasets, or research questions.

> 💡 **New to Python?**
>
> You do not need to understand every line of code immediately.
>
> Start by running an example, look at the result, and then experiment by changing one small thing at a time.

# Choose how you want to work

You can **read the book without installing anything**.

To run and modify the notebooks, choose the environment that best fits your needs.

# Run in your browser

For a quick start, use a browser-based Jupyter environment.

>- **Jupyter4NFDI** — run notebooks online using supported academic credentials

# What will we work with?

The training uses a real ioerDATA replication package on **climate regulation in cities**.

We will explore urban green infrastructure and related indicators while learning how research data moves from a repository into a reproducible Python workflow.

The focus is not only on getting the code to run — but also on understanding what the data means.

<hr style="border: none; border-top: 8px solid #70a401;">

# About this training

This training was developed by the **IOER Research Data Center (IOER RDC)** to provide practical guidance for working with research data from ioerDATA.

It is designed for students, researchers, practitioners, and anyone interested in learning reproducible spatial-data workflows with Python and Jupyter notebooks.

The training builds on the earlier training [Guide on Spatial and Biodiversity Data](https://training.fdz.ioer.info/intro.html).
