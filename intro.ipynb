```{admonition} Introduction
:class: note

Welcome to this hands-on course from [ioerDATA](https://data.fdz.ioer.de/) to insight! 🎉

**ioerDATA** is the institutional research‑data publication repository of the [Leibniz Institute of Ecological Urban and Regional Development (IOER)](https://www.ioer.de/en/).
In this course, you will learn how to work with real research data from ioerDATA using Python in a Jupyter Notebook environment provided by CartoLab.

The training is inspired by the teaching philosophy of the [IOER Research Data Center (IOER RDC)](https://ioer-fdz.de/en/): learning by doing, transparency in data handling, and reproducible research workflows.

This Jupyter notebook is designed to help students, researchers, scientists, and enthusiasts get started with data-driven research. Whether you're a beginner or an experienced data scientist, this book will help you fetch data programmatically, understand its structure and metadata, and build visualisations to derive insights from IOER’s geospatial datasets.
```

### Content
Our goal is to promote open science by providing access to the ioerDATA programmatically using its API (Application Programming Interface), which allows you to:
- **Download** and analyse data from our repository.
- **Explore** data programmatically using your preferred programming language (the examples in this course use Python).

ioerData is an open research data repository operated by the [Leibniz Institute of Ecological Urban and Regional Development (IOER)](https://www.ioer.de/). It provides access to datasets, codes, and documentation related to spatial, environmental, and socio-economic research. 

Our datasets are openly accessible and reusable, adhering to the **FAIR** (Findable, Accessible, Interoperable, Reusable) principles.


### Who can use this Jupyter Book?
This JupyterBook is suitable for:
- **Students, scientists and researchers** at various career stages.
- **Python beginners** who want to explore spatial and biodiversity data.
- **Experienced Python users** looking to refine their skills in these domains.


```{admonition} Tip
:class: hint, dropdown
A basic understanding of Python is beneficial but not required. Readers will need access to a computer with internet connectivity and the ability to run Jupyter Notebooks is recommended.
```

### How to use this book
You can use this Jupyter Book in different ways depending on your needs.

If you are new to the topic, follow the chapters in order. If you are working on a specific project, you can reuse individual explanations, code snippets, or complete workflows as building blocks for your own analysis.
