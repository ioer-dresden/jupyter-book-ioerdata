## Welcome!
This book explains how to access and work with research data from ioerDATA using Python.

It is designed for students and researchers who want to integrate geospatial data into their work — though anyone interested in reproducible, data-driven research may find it valuable.

The core content is promoting open science by teaching how to access, analyze, and reuse ioerDATA’s datasets via its API (Application Programming Interface), using Python in a Jupyter Notebook environment as the primary tool.

You’ll learn to retrieve, analyze, and visualize real-world datasets from ioerDATA, fostering skills for reproducible workflows.

**Publication Date:** August 2026
